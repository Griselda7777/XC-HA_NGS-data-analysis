```python
import pandas as pd
import numpy as np
import plotly.express as px



def enrichment_counts(pre_file, post_file):

	pre_file_read = pd.read_csv(pre_file)
	pre_count = np.array(pre_file_read['Counts'])
	pre_ratio = pre_count/sum(pre_count)

	post_file_read = pd.read_csv(post_file)
	post_count = np.array(post_file_read['Counts'])
	post_ratio = post_count/sum(post_count)
    
	try:
		enrichment_fold = post_ratio/pre_ratio
	except ZeroDivisionError:
		pass

	post_file_read['enrichment_fold'] = enrichment_fold
	cutoff_data=post_file_read.query("Counts > 0").copy()

	cutoff_data.to_csv('resultp20.txt',index=False,header=False)

enrichment_counts('Position_only_HA2_1.txt', 'Position_only_20_eval.txt')
# (pre, post-selection)
```
